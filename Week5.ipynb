```python
#program that counts the no of vowels and consonants in a string
a=str(input())
b=a.lower()
vowels="aeiou"
count_v=0;
count_c=0;
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("the no of vowels",count_v)
print("the no of consonants",count_c)


```

```python
#program that checks the sub string is in the original string if yes print the index of substring
s1=input("enter a string :")
s2=input("enter a substring :")
if s2 in s1:
  print(s1.find(s2))
else:
  print("not present")
```
