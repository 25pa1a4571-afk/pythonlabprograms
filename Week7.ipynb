```python
#tuple operation
t1=(1,2,3,4,5)
t2=(6,7,8,9,10)
print(type(t1))
print(t1+t2) #concatenation
print(t1*4) #multiple times print
print(len(t1)) #length of tuple
for i in t1:  #iteration
  print(i)
print(t1[::-1])  #slicing
print(t2[2:5])
print(t2[-4:-2:1])
if 20 in t1:
  print("yes")
else:
  print("no")
  #function
print(sum(t1))
print(max(t2))
#methods
print(t1.count(1))
print(t2.index(3))
t3=tuple(map(int,input().split()))
print(t3)



```
