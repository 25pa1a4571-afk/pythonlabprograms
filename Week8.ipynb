```python
file=open("data.txt","a")
file.writelines(["\nhow are you","\nwhat are u doing"])
file.close()
```

```python
with open("data.txt","r") as file:
  content=file.read(10)
  print(content)
  print(file.tell())
  file.seek(18)
  print(file.read())
  print(file.tell())
```
