```python
L1=[1,2,'g',"hii",45]
L2=[4.5,7,4,2,0,9]
print(L1) #print list elements
print(L1[3]) #print index value
print(L1[1:3]) #slicing
print(L2[: :-1]) #print elements in reverse order
print(L2[:8]) #slicing prints upto the index 8
print(L1[:6]) #slicing prints upto the index 6 from starting
print(L1[0:]) #slicing prints upto the last index
print(L1+L2) #adding two lists
print(L1*3) #repetition
print(L2[0:4:2]) #slicing prints upto the index 4 with difference 2 from starting
print(2 in L1) #checks 2 in l1 or not
print(9 not in L2)
for i in L2:
  print(i,end=" ") #iteration through a list
```
