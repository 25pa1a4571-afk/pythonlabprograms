```python
#to check whether a given no is palindrome or not
def Palindrome(num):
  original=num
  rev=0
  while(num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  if(rev==original):
    print("it is palindrome")
  else:
    print("not palindrome")
Palindrome(121)


```

```python
#to print reverse of a string
def rev_string(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
print(rev_string("nari"))
```

```python
#function that returns multiple values
def multiple_values(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_values(1,2,3,4,5,6)
print(f"sum is {res_sum},count is {res_count},avg is {res_avg}")
```
